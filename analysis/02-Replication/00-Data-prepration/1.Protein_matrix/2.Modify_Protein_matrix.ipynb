# Modiffy the protein matrix to proper format
## Steps:
1. Make all protein name to upper
2. Since all ALS pateints only have baseline visit, we only keep the baseline visit

```r
# load the proten matrix /protein_matrix.parquet
library(arrow)
protein_matrix <- read_parquet("protein_matrix.parquet")
```

```r
dim(protein_matrix)
```

```r
# Make all colname to upper, except the eid
colnames(protein_matrix) <- ifelse(colnames(protein_matrix) == "eid", colnames(protein_matrix), toupper(colnames(protein_matrix)))

```
